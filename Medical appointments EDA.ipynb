```python
#Creating new column Lead time bucket to compare No-show rates based on how many days each patient had between scheduled day and appointment day

df['lead_time_bucket'] = df['lead_time_days'].apply(lambda x: 
                                              'Same Day' if x == 0 
                                              else 'Short (1-3 Days)' if 1 <= x <=3
                                              else 'Within a week' if 4 <= x <=7
                                              else 'Long lead (8+ days)' if x >= 8
                                              else 'Unknown')
```

```python
df
```

```python
#Determining if wait time affects no show rate
df.groupby('lead_time_bucket').agg(total_appointments = ('lead_time_bucket', 'count'), no_show_rate = ('No-show', lambda x: (x == 'Yes').mean() * 100)).round(2).reset_index().sort_values(by = 'no_show_rate')
```

```python
#Determining if Age affects no show rate

df['age_bracket'] = df['Age'].apply(lambda x: 
                                              'Child (1 - 12)' if 0 <= x <= 12 
                                              else 'Teen (13 - 19)' if  13 <= x <= 19
                                              else 'Young Adult (20 - 39)' if 20 <= x <= 39
                                              else ' Adult (40 - 59)' if 40 <= x <= 59
                                              else 'Senior (60+)')

df.groupby('age_bracket').agg(total_appointments = ('lead_time_bucket', 'count'), no_show_rate = ('No-show', lambda x: (x == 'Yes').mean() * 100)).round(2).reset_index()
```

```python
#Do SMS reminders help
df['SMS_received'].unique()
df.groupby('SMS_received').agg(sms_type_count = ('No-show', lambda x: (x == 'Yes').count()), no_show_rate = ('No-show', lambda x: (x == 'Yes').mean() * 100)).round(2).reset_index()
```

```python
#WHich neighborhoods have the highest risk?

pd.set_option('display.max.rows', 9999999)
neighborhoods = df.groupby('Neighbourhood').agg(appointment_count = ('No-show', lambda x: (x == 'Yes').count()), no_show_rate = ('No-show', lambda x: (x == 'Yes').mean() * 100)).query('appointment_count >= 100').round(2).reset_index().nlargest(15, 'no_show_rate')
neighborhoods['risk_rank'] = neighborhoods['no_show_rate'].rank(ascending = False, method = 'dense')
neighborhoods
```

```python
#Risk scoring per patient

df = df.sort_values(['PatientId', 'AppointmentDay'])

df['prior_appointments'] = (df.groupby('PatientId').cumcount())

df['prior_no_shows'] = (df['No-show'].eq('Yes')
    .groupby(df['PatientId'])
    .cumsum()
    .shift(1)
    .fillna(0)
    .astype(int)
)

result = df[['PatientId', 'AppointmentID', 'AppointmentDay', 'No-show', 'prior_appointments', 'prior_no_shows']].sort_values(['PatientId', 'AppointmentDay'])

result
```

```python
#Creating a risk assessment table

df = df.sort_values(['PatientId', 'AppointmentDay', 'AppointmentID']).copy()

df['prior_appointments'] = (df.groupby('PatientId').cumcount())

df['no_show_flag'] = df['No-show'].eq('Yes').astype(int)

df['prior_no_shows'] = (df.groupby('PatientId')['no_show_flag']
      .cumsum()
      .groupby(df['PatientId'])
      .shift(1)
      .fillna(0)
      .astype(int)
)

df['prior_no_show_rate'] = (df['prior_no_shows'].div(df['prior_appointments'].replace(0, np.nan)).round(2))

conditions = [
    df['prior_appointments'].eq(0),

    (df['prior_no_show_rate'].ge(0.5)) |
    (df['lead_time_days'].ge(8)),

    (df['prior_no_show_rate'].ge(0.2)) |
    (df['lead_time_days'].between(4, 7))
]

choices = [
    'New Patient - Monitor',
    'High Risk',
    'Medium Risk'
]

df['risk_tier'] = np.select(conditions, choices, default='Low Risk')

high_risk_appointments = (
    df[
        
        [
            'PatientId',
            'AppointmentID',
            'AppointmentDay',
            'Neighbourhood',
            'lead_time_days',
            'prior_appointments',
            'prior_no_shows',
            'prior_no_show_rate',
            'risk_tier'
        ]
    ]
    .sort_values('AppointmentDay')
    
)

high_risk_appointments
```
